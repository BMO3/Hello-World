```python
import requests
import time
from IPython.display import Audio, display

# --- 1. THE WRITING DESK ---
# Type your script inside the triple quotes below.
my_script = """
Type your segment here.
"""

# --- 2. THE FAIL-SAFE ENGINE ---
def run_sias():
    if len(my_script.strip()) < 5:
        print("Error: Script is too short. Please type your segment.")
        return

    API_KEY = "7d185109-7cbf-4970-a5c6-0dce1fc6dd2e"
    VOICE_ID = "02f6b61f-5637-48b6-9d98-6f22e201c710" # Rafiki

    headers = {"X-API-KEY": API_KEY, "Content-Type": "application/json"}
    payload = {"script": [{"text": my_script, "voice_id": VOICE_ID}]}

    print("Step 1: Submitting to Wondercraft...")
    try:
        res = requests.post("https://api.wondercraft.ai/v1/podcast/scripted",
                            json=payload, headers=headers, timeout=10)

        if res.status_code != 200:
            print(f"Submission Failed: {res.status_code} - {res.text}")
            return

        job_id = res.json().get("job_id")
        print(f"Step 2: Job Created ({job_id}). Rendering...")

        # Step 3: Polling with a HARD LIMIT (15 attempts / 30 seconds)
        for attempt in range(15):
            time.sleep(2)
            status_res = requests.get(f"https://api.wondercraft.ai/v1/podcast/{job_id}",
                                      headers=headers, timeout=10)
            data = status_res.json()
            status = data.get("status")

            if status == "completed":
                print("Step 3: Render Complete! Playing audio...")
                display(Audio(url=data.get("download_url"), autoplay=True))
                return
            elif status == "failed":
                print("Error: Wondercraft server failed to render this audio.")
                return
            else:
                print(f"Still rendering... (Attempt {attempt + 1}/15)")

        print("Error: Operation timed out. The server is taking too long.")

    except Exception as e:
        print(f"System Error: {str(e)}")

# RUN IT
run_sias()
```
